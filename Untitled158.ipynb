```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import yaml
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Seller Network and Product Availability Analysis and Prediction"
)

CSV_PATH = os.path.join(
    BASE_DIR, "RS_Session_262_AU_869_A_to_B_i.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

H5_PATH = os.path.join(BASE_DIR, "model.h5")
PKL_PATH = os.path.join(BASE_DIR, "model.pkl")
YAML_PATH = os.path.join(BASE_DIR, "config.yaml")
JSON_PATH = os.path.join(BASE_DIR, "metadata.json")

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\n========== LOADING DATASET ==========")

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(f"Dataset not found: {CSV_PATH}")

df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())

# ============================================================
# 3. DATA CLEANING
# ============================================================

print("\n========== DATA PREPROCESSING ==========")

STATE_COL = "State/UT"
FEATURE_COL = "Number of Seller Points"
TARGET_COL = "Number of Product listed"

required_columns = [STATE_COL, FEATURE_COL, TARGET_COL]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

# Remove aggregate total rows
df = df[
    df[STATE_COL].astype(str).str.strip().str.lower() != "total"
].copy()

df = df[
    df["Sl. No."].astype(str).str.strip().str.lower() != "total"
].copy()

# Convert numerical columns
for col in [FEATURE_COL, TARGET_COL]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove rows with missing target
df = df.dropna(subset=[TARGET_COL])

# Remove invalid negative counts
df = df[df[TARGET_COL] >= 0]
df = df[
    df[FEATURE_COL].isna() | (df[FEATURE_COL] >= 0)
]

df = df.reset_index(drop=True)

if len(df) < 10:
    raise ValueError("Insufficient records for training.")

print("\nCleaned Dataset:")
print(df.head())

print("\nNumber of Records:", len(df))
print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 4. FEATURE SELECTION
# ============================================================

X = df[[FEATURE_COL]]
y = df[TARGET_COL]

print("\nFeature:", FEATURE_COL)
print("Target:", TARGET_COL)

# ============================================================
# 5. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=SEED
)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

# ============================================================
# 6. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        min_samples_leaf=2,
        random_state=SEED
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=4,
        min_samples_leaf=2,
        random_state=SEED,
        n_jobs=-1
    )
}

results = {}
trained_models = {}

print("\n========== TRAINING ML MODELS ==========")

for name, estimator in models.items():

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", estimator)
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    # Product counts cannot be negative
    predictions = np.maximum(predictions, 0)

    mae = mean_absolute_error(y_test, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(y_test, predictions)

    results[name] = {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2)
    }

    trained_models[name] = pipeline

    print(f"\n{name}")
    print("MAE:", round(mae, 4))
    print("RMSE:", round(rmse, 4))
    print("R2:", round(r2, 4))

# ============================================================
# 7. SELECT BEST ML MODEL
# ============================================================

# Lowest test RMSE is used for this experimental comparison.
# For rigorous model selection, use cross-validation
# on training data and reserve test data for final evaluation.

best_model_name = min(
    results,
    key=lambda name: results[name]["RMSE"]
)

best_model = trained_models[best_model_name]

print("\n========== BEST ML MODEL ==========")
print("Best Model:", best_model_name)

# ============================================================
# 8. SAVE PKL MODEL
# ============================================================

with open(PKL_PATH, "wb") as file:
    pickle.dump(best_model, file)

print("\nPKL model saved:", PKL_PATH)

# ============================================================
# 9. NEURAL NETWORK PREPROCESSING
# ============================================================

print("\n========== NEURAL NETWORK ==========")

nn_imputer = SimpleImputer(strategy="median")
nn_scaler = StandardScaler()

X_train_imputed = nn_imputer.fit_transform(X_train)
X_test_imputed = nn_imputer.transform(X_test)

X_train_scaled = nn_scaler.fit_transform(X_train_imputed)
X_test_scaled = nn_scaler.transform(X_test_imputed)

# Scale target for more stable neural-network training
target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

# ============================================================
# 10. BUILD NEURAL NETWORK
# ============================================================

nn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),

    tf.keras.layers.Dense(
        32,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        16,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        8,
        activation="relu"
    ),

    tf.keras.layers.Dense(1)
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

nn_model.summary()

# ============================================================
# 11. TRAIN NEURAL NETWORK
# ============================================================

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_scaled,
    y_train_scaled,
    epochs=300,
    batch_size=4,
    validation_split=0.2,
    callbacks=[early_stopping],
    verbose=0
)

print("\nNeural Network Training Completed")
print("Epochs Trained:", len(history.history["loss"]))

# ============================================================
# 12. EVALUATE NEURAL NETWORK
# ============================================================

nn_predictions_scaled = nn_model.predict(
    X_test_scaled,
    verbose=0
)

nn_predictions = target_scaler.inverse_transform(
    nn_predictions_scaled
).flatten()

nn_predictions = np.maximum(nn_predictions, 0)

nn_mae = mean_absolute_error(
    y_test, nn_predictions
)

nn_rmse = np.sqrt(
    mean_squared_error(y_test, nn_predictions)
)

nn_r2 = r2_score(
    y_test, nn_predictions
)

results["Neural Network"] = {
    "MAE": float(nn_mae),
    "RMSE": float(nn_rmse),
    "R2": float(nn_r2)
}

print("\nNeural Network Results:")
print("MAE:", round(nn_mae, 4))
print("RMSE:", round(nn_rmse, 4))
print("R2:", round(nn_r2, 4))

# ============================================================
# 13. SAVE H5 MODEL
# ============================================================

nn_model.save(H5_PATH)

print("\nH5 model saved:", H5_PATH)

# ============================================================
# 14. SAVE CONFIGURATION YAML
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based State-Wise Seller Network "
            "and Product Availability Analysis and Prediction"
        ),
        "version": "1.0",
        "task": "Regression",
        "description": (
            "Predict product listing counts "
            "using state-wise seller point counts."
        )
    },

    "dataset": {
        "file": CSV_PATH,
        "state_column": STATE_COL,
        "feature_columns": [FEATURE_COL],
        "target_column": TARGET_COL,
        "records_after_cleaning": int(len(df))
    },

    "training": {
        "test_size": 0.25,
        "random_state": SEED,
        "neural_network_epochs": 300,
        "batch_size": 4,
        "optimizer": "adam",
        "loss": "mse"
    },

    "models": {
        "traditional": list(models.keys()),
        "deep_learning": "Neural Network",
        "best_sklearn_model": best_model_name
    },

    "output": {
        "h5": H5_PATH,
        "pkl": PKL_PATH,
        "yaml": YAML_PATH,
        "json": JSON_PATH
    }
}

with open(YAML_PATH, "w", encoding="utf-8") as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML configuration saved:", YAML_PATH)

# ============================================================
# 15. SAVE METADATA JSON
# ============================================================

metadata = {
    "project_name": (
        "AI-Based State-Wise Seller Network "
        "and Product Availability Analysis and Prediction"
    ),

    "created_at": datetime.now().isoformat(),

    "dataset_information": {
        "total_records": int(len(df)),
        "total_columns": int(df.shape[1]),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "total_seller_points": int(df[FEATURE_COL].sum()),
        "total_product_listings": int(df[TARGET_COL].sum()),
        "features": [FEATURE_COL],
        "target": TARGET_COL
    },

    "model_information": {
        "best_sklearn_model": best_model_name,
        "models_evaluated": list(results.keys()),
        "evaluation_metrics": [
            "MAE",
            "RMSE",
            "R2"
        ]
    },

    "model_performance": results,

    "neural_network": {
        "epochs_trained": int(
            len(history.history["loss"])
        ),
        "final_training_loss": float(
            history.history["loss"][-1]
        ),
        "final_validation_loss": float(
            history.history["val_loss"][-1]
        )
    },

    "preprocessing": {
        "sklearn": "Median imputation inside saved pipeline",
        "neural_network": (
            "Median imputation, feature standardization "
            "and target standardization"
        ),
        "neural_network_feature_imputer_statistics": (
            nn_imputer.statistics_.tolist()
        ),
        "neural_network_feature_scaler_mean": (
            nn_scaler.mean_.tolist()
        ),
        "neural_network_feature_scaler_scale": (
            nn_scaler.scale_.tolist()
        ),
        "neural_network_target_scaler_mean": (
            target_scaler.mean_.tolist()
        ),
        "neural_network_target_scaler_scale": (
            target_scaler.scale_.tolist()
        )
    },

    "saved_files": {
        "h5": os.path.basename(H5_PATH),
        "pkl": os.path.basename(PKL_PATH),
        "yaml": os.path.basename(YAML_PATH),
        "json": os.path.basename(JSON_PATH)
    }
}

with open(JSON_PATH, "w", encoding="utf-8") as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nJSON metadata saved:", JSON_PATH)

# ============================================================
# 16. SAMPLE PREDICTION
# ============================================================

print("\n========== SAMPLE PREDICTION ==========")

sample = pd.DataFrame({
    FEATURE_COL: [5]
})

sample_prediction = best_model.predict(sample)[0]
sample_prediction = max(0, sample_prediction)

print("Seller Points:", 5)
print(
    "Predicted Product Listings:",
    round(sample_prediction)
)

# ============================================================
# 17. FINAL SUMMARY
# ============================================================

print("\n========== MODEL COMPARISON ==========")

comparison_df = pd.DataFrame(results).T
print(comparison_df.round(4))

print("\n========== FILES GENERATED ==========")

for path in [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]:
    if os.path.exists(path):
        print("[SAVED]", path)
    else:
        print("[MISSING]", path)

print("\nProject Execution Completed Successfully!")
```
