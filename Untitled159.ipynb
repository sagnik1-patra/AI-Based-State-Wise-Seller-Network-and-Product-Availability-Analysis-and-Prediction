```python
import os
import json
import pickle
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT PATHS
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Seller Network and Product Availability Analysis and Prediction"
)

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_262_AU_869_A_to_B_i.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\n========== LOADING DATASET ==========")

df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())
print("\nColumns:", df.columns.tolist())

STATE_COL = "State/UT"
FEATURE_COL = "Number of Seller Points"
TARGET_COL = "Number of Product listed"

required = [STATE_COL, FEATURE_COL, TARGET_COL]

for col in required:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

# ============================================================
# 3. DATA CLEANING
# ============================================================

print("\n========== DATA CLEANING ==========")

# Remove aggregate rows such as Total
df = df[
    ~df[STATE_COL]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["total", "grand total"])
].copy()

# Convert numeric columns
for col in [FEATURE_COL, TARGET_COL]:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )

# Remove invalid target values
df = df.dropna(subset=[TARGET_COL])
df = df[df[TARGET_COL] >= 0]

# Remove negative seller counts
df = df[
    df[FEATURE_COL].isna() |
    (df[FEATURE_COL] >= 0)
]

df = df.reset_index(drop=True)

print("\nCleaned Dataset:")
print(df.head())
print("\nDataset Shape:", df.shape)

if len(df) < 10:
    raise ValueError("Not enough data for model training.")

# ============================================================
# 4. FEATURES AND TARGET
# ============================================================

X = df[[FEATURE_COL]]
y = df[TARGET_COL]

# Keep state names for prediction CSV
states = df[STATE_COL]

# ============================================================
# 5. TRAIN TEST SPLIT
# ============================================================

indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.25,
    random_state=SEED
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

test_states = states.iloc[test_idx].reset_index(drop=True)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

# ============================================================
# 6. MACHINE LEARNING MODELS
# ============================================================

models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        min_samples_leaf=2,
        random_state=SEED
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=4,
        min_samples_leaf=2,
        random_state=SEED,
        n_jobs=-1
    )
}

results = []
predictions_dict = {}
trained_models = {}

print("\n========== MODEL TRAINING ==========")

for name, estimator in models.items():

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", estimator)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)
    y_pred = np.maximum(y_pred, 0)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "R2_Percentage": float(r2 * 100)
    })

    predictions_dict[name] = y_pred
    trained_models[name] = pipeline

    print(f"\n{name}")
    print("MAE:", round(mae, 4))
    print("RMSE:", round(rmse, 4))
    print("R2:", round(r2, 4))

# ============================================================
# 7. NEURAL NETWORK
# ============================================================

print("\n========== NEURAL NETWORK TRAINING ==========")

feature_imputer = SimpleImputer(strategy="median")
feature_scaler = StandardScaler()
target_scaler = StandardScaler()

X_train_imp = feature_imputer.fit_transform(X_train)
X_test_imp = feature_imputer.transform(X_test)

X_train_scaled = feature_scaler.fit_transform(X_train_imp)
X_test_scaled = feature_scaler.transform(X_test_imp)

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

nn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(1)
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_scaled,
    y_train_scaled,
    epochs=300,
    batch_size=4,
    validation_split=0.2,
    callbacks=[early_stopping],
    verbose=0
)

nn_pred_scaled = nn_model.predict(
    X_test_scaled,
    verbose=0
)

nn_pred = target_scaler.inverse_transform(
    nn_pred_scaled
).flatten()

nn_pred = np.maximum(nn_pred, 0)

nn_mae = mean_absolute_error(y_test, nn_pred)
nn_rmse = np.sqrt(mean_squared_error(y_test, nn_pred))
nn_r2 = r2_score(y_test, nn_pred)

results.append({
    "Model": "Neural Network",
    "MAE": float(nn_mae),
    "RMSE": float(nn_rmse),
    "R2": float(nn_r2),
    "R2_Percentage": float(nn_r2 * 100)
})

predictions_dict["Neural Network"] = nn_pred

print("\nNeural Network:")
print("MAE:", round(nn_mae, 4))
print("RMSE:", round(nn_rmse, 4))
print("R2:", round(nn_r2, 4))

# ============================================================
# 8. RESULT CSV
# ============================================================

result_df = pd.DataFrame(results)

result_df = result_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

result_path = os.path.join(BASE_DIR, "result.csv")
result_df.to_csv(result_path, index=False)

print("\n========== RESULT CSV ==========")
print(result_df.to_string(index=False))
print("Saved:", result_path)

# ============================================================
# 9. PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "State/UT": test_states,
    "Seller_Points": X_test[FEATURE_COL].to_numpy(),
    "Actual_Product_Listings": y_test.to_numpy()
})

for name, pred in predictions_dict.items():
    column_name = name.replace(" ", "_") + "_Prediction"
    prediction_df[column_name] = np.round(pred, 2)

prediction_df = prediction_df.sort_values(
    by="Actual_Product_Listings",
    ascending=False
).reset_index(drop=True)

prediction_path = os.path.join(BASE_DIR, "prediction.csv")
prediction_df.to_csv(prediction_path, index=False)

print("\n========== PREDICTION CSV ==========")
print(prediction_df.to_string(index=False))
print("Saved:", prediction_path)

# ============================================================
# 10. SAVE BEST SKLEARN MODEL
# ============================================================

sklearn_results = result_df[
    result_df["Model"].isin(models.keys())
]

best_sklearn_name = sklearn_results.iloc[0]["Model"]
best_sklearn_model = trained_models[best_sklearn_name]

pkl_path = os.path.join(BASE_DIR, "model.pkl")

with open(pkl_path, "wb") as f:
    pickle.dump(best_sklearn_model, f)

print("\nBest sklearn model:", best_sklearn_name)
print("Saved:", pkl_path)

# ============================================================
# 11. SAVE NEURAL NETWORK H5
# ============================================================

h5_path = os.path.join(BASE_DIR, "model.h5")
nn_model.save(h5_path)

print("Saved:", h5_path)

# ============================================================
# 12. ACCURACY GRAPH (R2 SCORE)
# ============================================================

print("\n========== ACCURACY GRAPH ==========")

plt.figure(figsize=(11, 6))

colors = sns.color_palette(
    "viridis",
    len(result_df)
)

bars = plt.bar(
    result_df["Model"],
    result_df["R2_Percentage"],
    color=colors
)

plt.axhline(0, color="black", linewidth=1)

for bar, value in zip(
    bars,
    result_df["R2_Percentage"]
):
    plt.annotate(
        f"{value:.2f}%",
        (
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        ha="center",
        va="bottom" if value >= 0 else "top",
        fontsize=10
    )

plt.title(
    "Model Performance - R2 Score",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R2 Score (%)")
plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

accuracy_path = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

plt.savefig(
    accuracy_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", accuracy_path)

# ============================================================
# 13. PREDICTION GRAPH
# ============================================================

print("\n========== PREDICTION GRAPH ==========")

best_model_name = result_df.iloc[0]["Model"]

best_prediction_col = (
    best_model_name.replace(" ", "_") + "_Prediction"
)

plt.figure(figsize=(13, 7))

x_axis = np.arange(len(prediction_df))

plt.plot(
    x_axis,
    prediction_df["Actual_Product_Listings"],
    marker="o",
    linewidth=2,
    label="Actual Product Listings"
)

plt.plot(
    x_axis,
    prediction_df[best_prediction_col],
    marker="s",
    linestyle="--",
    linewidth=2,
    label=f"Predicted - {best_model_name}"
)

plt.title(
    "Actual vs Predicted Product Listings",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Number of Product Listings")

plt.xticks(
    x_axis,
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

prediction_graph_path = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)

plt.savefig(
    prediction_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", prediction_graph_path)

# ============================================================
# 14. RESULT GRAPH
# ============================================================

print("\n========== RESULT GRAPH ==========")

x_axis = np.arange(len(result_df))
width = 0.35

plt.figure(figsize=(12, 7))

plt.bar(
    x_axis - width / 2,
    result_df["MAE"],
    width,
    label="MAE"
)

plt.bar(
    x_axis + width / 2,
    result_df["RMSE"],
    width,
    label="RMSE"
)

plt.title(
    "Machine Learning Model Error Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Error (Product Listings)")

plt.xticks(
    x_axis,
    result_df["Model"],
    rotation=15
)

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

result_graph_path = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

plt.savefig(
    result_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", result_graph_path)

# ============================================================
# 15. COMPARISON GRAPH
# ============================================================

print("\n========== COMPARISON GRAPH ==========")

plt.figure(figsize=(14, 8))

plt.plot(
    prediction_df["Actual_Product_Listings"].to_numpy(),
    marker="o",
    linewidth=3,
    label="Actual Values",
    color="black"
)

for name in predictions_dict.keys():

    column_name = name.replace(" ", "_") + "_Prediction"

    plt.plot(
        prediction_df[column_name].to_numpy(),
        marker=".",
        linestyle="--",
        linewidth=1.8,
        label=name
    )

plt.title(
    "Comparison of All ML Model Predictions",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Test States")
plt.ylabel("Number of Product Listings")

plt.xticks(
    np.arange(len(prediction_df)),
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

comparison_path = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

plt.savefig(
    comparison_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", comparison_path)

# ============================================================
# 16. CORRELATION HEATMAP
# ============================================================

print("\n========== HEATMAP ==========")

heatmap_df = df[
    [FEATURE_COL, TARGET_COL]
].copy()

# Include a meaningful derived metric
heatmap_df["Products_Per_Seller"] = np.where(
    heatmap_df[FEATURE_COL] > 0,
    heatmap_df[TARGET_COL] /
    heatmap_df[FEATURE_COL],
    np.nan
)

correlation = heatmap_df.corr(
    method="pearson"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0,
    square=True
)

plt.title(
    "Correlation Heatmap - Seller Network Dataset",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

heatmap_path = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

plt.savefig(
    heatmap_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", heatmap_path)

# ============================================================
# 17. SAVE YAML CONFIG
# ============================================================

config = {
    "project_name": (
        "AI-Based State-Wise Seller Network "
        "and Product Availability Analysis and Prediction"
    ),
    "dataset": CSV_PATH,
    "task": "Regression",
    "features": [FEATURE_COL],
    "target": TARGET_COL,
    "test_size": 0.25,
    "random_state": SEED,
    "models": list(predictions_dict.keys()),
    "best_sklearn_model": best_sklearn_name,
    "best_test_rmse_model": best_model_name,
    "output_directory": BASE_DIR
}

yaml_path = os.path.join(
    BASE_DIR,
    "config.yaml"
)

with open(yaml_path, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", yaml_path)

# ============================================================
# 18. SAVE JSON METADATA
# ============================================================

metadata = {
    "project": config["project_name"],
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),
    "total_seller_points": float(df[FEATURE_COL].sum()),
    "total_product_listings": float(df[TARGET_COL].sum()),
    "feature": FEATURE_COL,
    "target": TARGET_COL,
    "best_sklearn_model": best_sklearn_name,
    "best_test_rmse_model": best_model_name,
    "results": result_df.to_dict(orient="records"),
    "neural_network_preprocessing": {
        "imputer_statistics": feature_imputer.statistics_.tolist(),
        "feature_mean": feature_scaler.mean_.tolist(),
        "feature_scale": feature_scaler.scale_.tolist(),
        "target_mean": target_scaler.mean_.tolist(),
        "target_scale": target_scaler.scale_.tolist()
    },
    "generated_files": [
        "accuracy_graph.png",
        "prediction.csv",
        "prediction_graph.png",
        "result.csv",
        "result_graph.png",
        "comparison_graph.png",
        "heatmap.png",
        "model.pkl",
        "model.h5",
        "config.yaml",
        "metadata.json"
    ]
}

json_path = os.path.join(
    BASE_DIR,
    "metadata.json"
)

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(
        metadata,
        f,
        indent=4,
        ensure_ascii=False
    )

print("Saved:", json_path)

# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n======================================")
print("PROJECT EXECUTION COMPLETED")
print("======================================")

print("\nMODEL PERFORMANCE:")
print(result_df.to_string(index=False))

print("\nBEST MODEL BY TEST RMSE:", best_model_name)

print("\nGENERATED FILES:")

for filename in metadata["generated_files"]:
    path = os.path.join(BASE_DIR, filename)

    if os.path.exists(path):
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nAll graphs displayed and saved successfully.")
print("\nOutput Directory:")
print(BASE_DIR)
```
