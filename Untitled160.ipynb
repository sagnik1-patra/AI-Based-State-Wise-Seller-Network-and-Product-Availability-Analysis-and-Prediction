```python

import os
import json
import pickle
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import tensorflow as tf

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Seller Network and Product Availability Analysis and Prediction"
)

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_262_AU_869_A_to_B_i.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

STATE_COL = "State/UT"
FEATURE_COL = "Number of Seller Points"
TARGET_COL = "Number of Product listed"

# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def output_path(filename):
    return os.path.join(BASE_DIR, "ais_" + filename)


def save_show(filename):
    path = output_path(filename)
    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )
    print("Saved:", path)
    plt.show()
    plt.close()


def calculate_metrics(actual, predicted):
    predicted = np.maximum(
        np.asarray(predicted).flatten(),
        0
    )

    actual = np.asarray(actual).flatten()

    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )
    r2 = r2_score(actual, predicted)

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "R2_Percentage": float(r2 * 100)
    }


# ============================================================
# 3. LOAD AND CLEAN DATASET
# ============================================================

print("\n========== LOADING DATASET ==========")

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(CSV_PATH)

df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)

df.columns = df.columns.str.strip()

required = [
    STATE_COL,
    FEATURE_COL,
    TARGET_COL
]

for col in required:
    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

# Remove total rows
df = df[
    ~df[STATE_COL]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["total", "grand total"])
].copy()

if "Sl. No." in df.columns:
    df = df[
        ~df["Sl. No."]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(["total", "grand total"])
    ].copy()

# Convert numeric columns
for col in [FEATURE_COL, TARGET_COL]:
    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

df = df.dropna(subset=[TARGET_COL])
df = df[df[TARGET_COL] >= 0]

df = df[
    df[FEATURE_COL].isna() |
    (df[FEATURE_COL] >= 0)
]

df = df.reset_index(drop=True)

if len(df) < 12:
    raise ValueError(
        "Insufficient data for AIS training."
    )

print("\nCleaned Dataset:")
print(df.head())
print("\nRecords:", len(df))

# ============================================================
# 4. FEATURE AND TARGET
# ============================================================

X = df[[FEATURE_COL]]
y = df[TARGET_COL]

indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.25,
    random_state=SEED
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

test_states = df.iloc[test_idx][
    STATE_COL
].reset_index(drop=True)

print("\nTraining:", len(X_train))
print("Testing:", len(X_test))

# ============================================================
# 5. AIS - CLONAL SELECTION ALGORITHM
# ============================================================

print("\n========== AIS OPTIMIZATION ==========")

# Antibody represents Random Forest hyperparameters:
# [n_estimators, max_depth, min_samples_leaf,
#  min_samples_split, max_features_fraction]

BOUNDS = [
    (30, 180),
    (2, 12),
    (1, 5),
    (2, 10),
    (0.5, 1.0)
]

INTEGER_INDICES = {0, 1, 2, 3}

POPULATION_SIZE = 8
GENERATIONS = 12
CLONES_PER_PARENT = 4
MUTATION_RATE = 0.25
RANDOM_REPLACEMENTS = 2


def random_antibody():
    antibody = []

    for i, (low, high) in enumerate(BOUNDS):
        if i in INTEGER_INDICES:
            antibody.append(
                random.randint(int(low), int(high))
            )
        else:
            antibody.append(
                random.uniform(low, high)
            )

    return antibody


def repair_antibody(antibody):
    repaired = []

    for i, value in enumerate(antibody):
        low, high = BOUNDS[i]

        value = np.clip(value, low, high)

        if i in INTEGER_INDICES:
            value = int(round(value))
        else:
            value = float(value)

        repaired.append(value)

    return repaired


def antibody_to_params(antibody):
    a = repair_antibody(antibody)

    return {
        "n_estimators": a[0],
        "max_depth": a[1],
        "min_samples_leaf": a[2],
        "min_samples_split": a[3],
        "max_features": a[4]
    }


def create_rf_pipeline(params):
    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=1
            )
        )
    ])


cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}


def fitness(antibody):
    params = antibody_to_params(antibody)

    key = tuple(params.items())

    if key in fitness_cache:
        return fitness_cache[key]

    model = create_rf_pipeline(params)

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    rmse = np.sqrt(
        np.maximum(-scores, 0)
    ).mean()

    fitness_cache[key] = float(rmse)

    return float(rmse)


def mutate_antibody(antibody, rate):
    mutated = list(antibody)

    for i in range(len(mutated)):
        if random.random() < rate:
            low, high = BOUNDS[i]
            span = high - low

            mutated[i] += random.gauss(
                0,
                0.15 * span
            )

    return repair_antibody(mutated)


def clonal_selection():
    population = [
        random_antibody()
        for _ in range(POPULATION_SIZE)
    ]

    history = []

    best_antibody = None
    best_fitness = float("inf")

    for generation in range(1, GENERATIONS + 1):

        ranked = sorted(
            population,
            key=fitness
        )

        current_best = ranked[0]
        current_fitness = fitness(current_best)

        if current_fitness < best_fitness:
            best_fitness = current_fitness
            best_antibody = list(current_best)

        # Keep the strongest antibodies
        selected = ranked[:POPULATION_SIZE // 2]

        clones = []

        for rank, antibody in enumerate(selected):

            parent_fitness = fitness(antibody)

            # Stronger antibodies receive lower mutation
            adaptive_rate = min(
                0.8,
                MUTATION_RATE *
                (1 + rank / len(selected))
            )

            for _ in range(CLONES_PER_PARENT):
                clone = mutate_antibody(
                    antibody,
                    adaptive_rate
                )
                clones.append(clone)

        candidates = (
            ranked +
            clones +
            [
                random_antibody()
                for _ in range(RANDOM_REPLACEMENTS)
            ]
        )

        candidates = sorted(
            candidates,
            key=fitness
        )

        # Elitism: retain global best
        candidates.append(best_antibody)

        population = sorted(
            candidates,
            key=fitness
        )[:POPULATION_SIZE]

        history.append({
            "Generation": generation,
            "Best_CV_RMSE": best_fitness,
            "Generation_Best_RMSE": current_fitness
        })

        print(
            f"Generation {generation}/{GENERATIONS} "
            f"| Best CV RMSE: {best_fitness:.4f}"
        )

    return (
        best_antibody,
        best_fitness,
        pd.DataFrame(history)
    )


best_antibody, best_cv_rmse, history_df = (
    clonal_selection()
)

best_params = antibody_to_params(best_antibody)

print("\nBest AIS Parameters:")
print(best_params)

print("Best CV RMSE:", best_cv_rmse)

# ============================================================
# 6. TRAIN AIS RANDOM FOREST
# ============================================================

print("\n========== AIS RANDOM FOREST ==========")

ais_rf = create_rf_pipeline(best_params)
ais_rf.fit(X_train, y_train)

rf_pred = ais_rf.predict(X_test)
rf_pred = np.maximum(rf_pred, 0)

rf_metrics = calculate_metrics(
    y_test,
    rf_pred
)

print("AIS Random Forest Metrics:")
print(rf_metrics)

# ============================================================
# 7. SAVE AIS PKL MODEL
# ============================================================

with open(
    output_path("model.pkl"),
    "wb"
) as f:
    pickle.dump(ais_rf, f)

print("Saved AIS PKL model")

# ============================================================
# 8. AIS NEURAL NETWORK OPTIMIZATION
# ============================================================

print("\n========== AIS NEURAL NETWORK ==========")

# Optimize neural network hyperparameters with AIS.
# Each antibody: [units1, units2, learning_rate]

NN_BOUNDS = [
    (8, 64),
    (4, 32),
    (0.0003, 0.01)
]

NN_POPULATION = 5
NN_GENERATIONS = 5
NN_CLONES = 2


def random_nn_antibody():
    return [
        random.randint(8, 64),
        random.randint(4, 32),
        random.uniform(0.0003, 0.01)
    ]


def repair_nn_antibody(a):
    return [
        int(np.clip(round(a[0]), 8, 64)),
        int(np.clip(round(a[1]), 4, 32)),
        float(np.clip(a[2], 0.0003, 0.01))
    ]


def mutate_nn_antibody(a):
    new = list(a)

    if random.random() < 0.5:
        new[0] += random.randint(-12, 12)

    if random.random() < 0.5:
        new[1] += random.randint(-8, 8)

    if random.random() < 0.5:
        new[2] *= random.uniform(0.5, 1.5)

    return repair_nn_antibody(new)


def build_nn(params):
    units1, units2, lr = params

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(1,)),
        tf.keras.layers.Dense(
            units1,
            activation="relu"
        ),
        tf.keras.layers.Dense(
            units2,
            activation="relu"
        ),
        tf.keras.layers.Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=lr
        ),
        loss="mse",
        metrics=["mae"]
    )

    return model


# Split training data again for NN optimization.
# Test data stays untouched.

nn_train_idx, nn_val_idx = train_test_split(
    np.arange(len(X_train)),
    test_size=0.25,
    random_state=SEED
)

X_nn_train = X_train.iloc[nn_train_idx]
X_nn_val = X_train.iloc[nn_val_idx]

y_nn_train = y_train.iloc[nn_train_idx]
y_nn_val = y_train.iloc[nn_val_idx]

nn_search_imputer = SimpleImputer(
    strategy="median"
)

nn_search_scaler = StandardScaler()
nn_search_target_scaler = StandardScaler()

X_nn_train_imp = nn_search_imputer.fit_transform(
    X_nn_train
)

X_nn_val_imp = nn_search_imputer.transform(
    X_nn_val
)

X_nn_train_scaled = nn_search_scaler.fit_transform(
    X_nn_train_imp
)

X_nn_val_scaled = nn_search_scaler.transform(
    X_nn_val_imp
)

y_nn_train_scaled = nn_search_target_scaler.fit_transform(
    y_nn_train.to_numpy().reshape(-1, 1)
)

nn_fitness_cache = {}


def nn_fitness(antibody):
    params = repair_nn_antibody(antibody)
    key = (
        params[0],
        params[1],
        round(params[2], 7)
    )

    if key in nn_fitness_cache:
        return nn_fitness_cache[key]

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(SEED)

    model = build_nn(params)

    stopper = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=12,
        restore_best_weights=True
    )

    model.fit(
        X_nn_train_scaled,
        y_nn_train_scaled,
        validation_data=(
            X_nn_val_scaled,
            nn_search_target_scaler.transform(
                y_nn_val.to_numpy().reshape(-1, 1)
            )
        ),
        epochs=100,
        batch_size=4,
        callbacks=[stopper],
        verbose=0
    )

    pred_scaled = model.predict(
        X_nn_val_scaled,
        verbose=0
    )

    pred = nn_search_target_scaler.inverse_transform(
        pred_scaled
    ).flatten()

    pred = np.maximum(pred, 0)

    score = np.sqrt(
        mean_squared_error(y_nn_val, pred)
    )

    nn_fitness_cache[key] = float(score)

    return float(score)


def optimize_nn():
    population = [
        random_nn_antibody()
        for _ in range(NN_POPULATION)
    ]

    best = None
    best_score = float("inf")

    nn_history = []

    for gen in range(1, NN_GENERATIONS + 1):

        ranked = sorted(
            population,
            key=nn_fitness
        )

        if nn_fitness(ranked[0]) < best_score:
            best = list(ranked[0])
            best_score = nn_fitness(best)

        selected = ranked[:3]

        clones = []

        for parent in selected:
            for _ in range(NN_CLONES):
                clones.append(
                    mutate_nn_antibody(parent)
                )

        candidates = (
            ranked +
            clones +
            [random_nn_antibody(), best]
        )

        population = sorted(
            candidates,
            key=nn_fitness
        )[:NN_POPULATION]

        nn_history.append({
            "Generation": gen,
            "Best_Validation_RMSE": best_score
        })

        print(
            f"NN Generation {gen}/{NN_GENERATIONS} "
            f"| Best RMSE: {best_score:.4f}"
        )

    return (
        repair_nn_antibody(best),
        best_score,
        pd.DataFrame(nn_history)
    )


best_nn_params, best_nn_score, nn_history_df = (
    optimize_nn()
)

print("\nBest NN Parameters:", best_nn_params)

# ============================================================
# 9. TRAIN FINAL AIS NEURAL NETWORK
# ============================================================

final_imputer = SimpleImputer(
    strategy="median"
)

final_feature_scaler = StandardScaler()
final_target_scaler = StandardScaler()

X_train_imp = final_imputer.fit_transform(X_train)
X_test_imp = final_imputer.transform(X_test)

X_train_scaled = final_feature_scaler.fit_transform(
    X_train_imp
)

X_test_scaled = final_feature_scaler.transform(
    X_test_imp
)

y_train_scaled = final_target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

ais_nn = build_nn(best_nn_params)

stopper = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

nn_history_final = ais_nn.fit(
    X_train_scaled,
    y_train_scaled,
    validation_split=0.2,
    epochs=250,
    batch_size=4,
    callbacks=[stopper],
    verbose=0
)

nn_pred_scaled = ais_nn.predict(
    X_test_scaled,
    verbose=0
)

nn_pred = final_target_scaler.inverse_transform(
    nn_pred_scaled
).flatten()

nn_pred = np.maximum(nn_pred, 0)

nn_metrics = calculate_metrics(
    y_test,
    nn_pred
)

print("\nAIS Neural Network Metrics:")
print(nn_metrics)

ais_nn.save(
    output_path("model.h5")
)

print("Saved AIS H5 model")

# ============================================================
# 10. RESULT CSV
# ============================================================

result_df = pd.DataFrame([
    {
        "Model": "AIS Random Forest",
        **rf_metrics
    },
    {
        "Model": "AIS Neural Network",
        **nn_metrics
    }
])

result_df = result_df.sort_values(
    "RMSE"
).reset_index(drop=True)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\n========== AIS RESULTS ==========")
print(result_df.to_string(index=False))

# ============================================================
# 11. PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "State/UT": test_states,
    "Seller_Points": X_test[FEATURE_COL].to_numpy(),
    "Actual_Product_Listings": y_test.to_numpy(),
    "AIS_RF_Prediction": np.round(rf_pred, 2),
    "AIS_NN_Prediction": np.round(nn_pred, 2)
})

prediction_df = prediction_df.sort_values(
    "Actual_Product_Listings",
    ascending=False
).reset_index(drop=True)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\n========== AIS PREDICTIONS ==========")
print(prediction_df.to_string(index=False))

# ============================================================
# 12. OPTIMIZATION HISTORY CSV
# ============================================================

history_df.to_csv(
    output_path("optimization_history.csv"),
    index=False
)

nn_history_df.to_csv(
    output_path("nn_optimization_history.csv"),
    index=False
)

# ============================================================
# 13. AIS ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

bars = plt.bar(
    result_df["Model"],
    result_df["R2_Percentage"],
    color=["#2878B5", "#E69F00"]
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

for bar, value in zip(
    bars,
    result_df["R2_Percentage"]
):
    plt.annotate(
        f"{value:.2f}%",
        (
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        ha="center",
        va="bottom" if value >= 0 else "top"
    )

plt.title(
    "AIS Model Performance - R2 Score",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("R2 Score (%)")
plt.xlabel("AIS Models")
plt.grid(axis="y", alpha=0.3)

save_show("accuracy_graph.png")

# ============================================================
# 14. AIS PREDICTION GRAPH
# ============================================================

best_model_name = result_df.iloc[0]["Model"]

best_prediction_col = (
    "AIS_RF_Prediction"
    if best_model_name == "AIS Random Forest"
    else "AIS_NN_Prediction"
)

x = np.arange(len(prediction_df))

plt.figure(figsize=(13, 7))

plt.plot(
    x,
    prediction_df["Actual_Product_Listings"],
    marker="o",
    linewidth=2.5,
    label="Actual"
)

plt.plot(
    x,
    prediction_df[best_prediction_col],
    marker="s",
    linestyle="--",
    linewidth=2,
    label=best_model_name
)

plt.xticks(
    x,
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.title(
    "AIS Actual vs Predicted Product Listings",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / UT")
plt.ylabel("Product Listings")
plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ============================================================
# 15. AIS RESULT GRAPH
# ============================================================

x = np.arange(len(result_df))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width / 2,
    result_df["MAE"],
    width,
    label="MAE"
)

plt.bar(
    x + width / 2,
    result_df["RMSE"],
    width,
    label="RMSE"
)

plt.xticks(
    x,
    result_df["Model"]
)

plt.title(
    "AIS Model Error Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("Error (Product Listings)")
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 16. AIS COMPARISON GRAPH
# ============================================================

x = np.arange(len(prediction_df))

plt.figure(figsize=(14, 7))

plt.plot(
    x,
    prediction_df["Actual_Product_Listings"],
    color="black",
    marker="o",
    linewidth=3,
    label="Actual"
)

plt.plot(
    x,
    prediction_df["AIS_RF_Prediction"],
    marker="s",
    linestyle="--",
    label="AIS Random Forest"
)

plt.plot(
    x,
    prediction_df["AIS_NN_Prediction"],
    marker="^",
    linestyle="--",
    label="AIS Neural Network"
)

plt.xticks(
    x,
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.title(
    "AIS Model Prediction Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / UT")
plt.ylabel("Product Listings")
plt.legend()
plt.grid(alpha=0.3)

save_show("comparison_graph.png")

# ============================================================
# 17. AIS HEATMAP
# ============================================================

heatmap_df = df[
    [FEATURE_COL, TARGET_COL]
].copy()

heatmap_df["Products_Per_Seller"] = np.where(
    heatmap_df[FEATURE_COL] > 0,
    heatmap_df[TARGET_COL] /
    heatmap_df[FEATURE_COL],
    np.nan
)

corr = heatmap_df.corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0,
    square=True
)

plt.title(
    "AIS Project - Dataset Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

save_show("heatmap.png")

# ============================================================
# 18. AIS FITNESS GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Generation"],
    history_df["Best_CV_RMSE"],
    marker="o",
    linewidth=2.5,
    label="AIS Random Forest"
)

plt.xlabel("Generation")
plt.ylabel("Best CV RMSE")

plt.title(
    "AIS Clonal Selection Optimization Progress",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.3)

save_show("fitness_graph.png")

# ============================================================
# 19. AIS NN FITNESS GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    nn_history_df["Generation"],
    nn_history_df["Best_Validation_RMSE"],
    marker="o",
    linewidth=2.5,
    label="AIS Neural Network"
)

plt.title(
    "AIS Neural Network Optimization Progress",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Best Validation RMSE")
plt.legend()
plt.grid(alpha=0.3)

save_show("nn_fitness_graph.png")

# ============================================================
# 20. SAVE YAML
# ============================================================

config = {
    "project": (
        "AI-Based State-Wise Seller Network "
        "and Product Availability Analysis and Prediction"
    ),
    "algorithm": "Artificial Immune System",
    "optimization_method": "Clonal Selection",
    "dataset": CSV_PATH,
    "feature": FEATURE_COL,
    "target": TARGET_COL,
    "random_state": SEED,
    "test_size": 0.25,
    "ais_random_forest": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "clones_per_parent": CLONES_PER_PARENT,
        "mutation_rate": MUTATION_RATE,
        "best_hyperparameters": best_params,
        "best_cv_rmse": float(best_cv_rmse)
    },
    "ais_neural_network": {
        "population_size": NN_POPULATION,
        "generations": NN_GENERATIONS,
        "clones_per_parent": NN_CLONES,
        "best_parameters": {
            "units1": best_nn_params[0],
            "units2": best_nn_params[1],
            "learning_rate": best_nn_params[2]
        },
        "best_validation_rmse": float(best_nn_score)
    },
    "output_directory": BASE_DIR
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

# ============================================================
# 21. SAVE JSON METADATA
# ============================================================

metadata = {
    "project": config["project"],
    "algorithm": "AIS - Clonal Selection",
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),
    "total_seller_points": float(
        df[FEATURE_COL].sum()
    ),
    "total_product_listings": float(
        df[TARGET_COL].sum()
    ),
    "ais_rf_best_parameters": best_params,
    "ais_rf_best_cv_rmse": float(best_cv_rmse),
    "ais_nn_best_parameters": {
        "units1": best_nn_params[0],
        "units2": best_nn_params[1],
        "learning_rate": best_nn_params[2]
    },
    "ais_nn_best_validation_rmse": float(
        best_nn_score
    ),
    "model_results": result_df.to_dict(
        orient="records"
    ),
    "neural_network_preprocessing": {
        "imputer_statistics": (
            final_imputer.statistics_.tolist()
        ),
        "feature_mean": (
            final_feature_scaler.mean_.tolist()
        ),
        "feature_scale": (
            final_feature_scaler.scale_.tolist()
        ),
        "target_mean": (
            final_target_scaler.mean_.tolist()
        ),
        "target_scale": (
            final_target_scaler.scale_.tolist()
        )
    },
    "files": [
        "ais_model.pkl",
        "ais_model.h5",
        "ais_config.yaml",
        "ais_metadata.json",
        "ais_accuracy_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_comparison_graph.png",
        "ais_heatmap.png",
        "ais_fitness_graph.png",
        "ais_nn_fitness_graph.png",
        "ais_optimization_history.csv",
        "ais_nn_optimization_history.csv"
    ]
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4,
        ensure_ascii=False
    )

# ============================================================
# 22. FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("==========================================")

print("\nAIS Random Forest:")
print(rf_metrics)

print("\nAIS Neural Network:")
print(nn_metrics)

print("\nBest Model by Test RMSE:")
print(best_model_name)

print("\nGenerated Files:")

for filename in metadata["files"]:
    path = os.path.join(BASE_DIR, filename)

    if os.path.exists(path):
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOutput Directory:")
print(BASE_DIR)
```
