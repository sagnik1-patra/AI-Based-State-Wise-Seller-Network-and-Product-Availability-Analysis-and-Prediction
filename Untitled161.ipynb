```python

import os
import json
import pickle
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import tensorflow as tf

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Seller Network and Product Availability Analysis and Prediction"
)

CSV_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_262_AU_869_A_to_B_i.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)

sns.set_theme(style="whitegrid")

STATE_COL = "State/UT"
FEATURE_COL = "Number of Seller Points"
TARGET_COL = "Number of Product listed"

# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def output_path(filename):
    return os.path.join(
        BASE_DIR,
        "pso_" + filename
    )


def save_show(filename):
    path = output_path(filename)

    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print("Saved:", path)

    plt.show()
    plt.close()


def calculate_metrics(actual, predicted):

    actual = np.asarray(actual).flatten()

    predicted = np.maximum(
        np.asarray(predicted).flatten(),
        0
    )

    mae = mean_absolute_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )

    r2 = r2_score(
        actual,
        predicted
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "R2_Percentage": float(r2 * 100)
    }


# ============================================================
# 3. LOAD DATASET
# ============================================================

print("\n========== LOADING DATASET ==========")

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(
    CSV_PATH,
    encoding="utf-8-sig"
)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nColumns:")
print(df.columns.tolist())

required = [
    STATE_COL,
    FEATURE_COL,
    TARGET_COL
]

for col in required:
    if col not in df.columns:
        raise ValueError(
            f"Missing required column: {col}"
        )

# ============================================================
# 4. DATA CLEANING
# ============================================================

print("\n========== DATA CLEANING ==========")

# Remove aggregate total rows
df = df[
    ~df[STATE_COL]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["total", "grand total"])
].copy()

if "Sl. No." in df.columns:
    df = df[
        ~df["Sl. No."]
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(["total", "grand total"])
    ].copy()

# Convert numeric columns
for col in [FEATURE_COL, TARGET_COL]:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

# Remove invalid records
df = df.dropna(
    subset=[TARGET_COL]
)

df = df[
    df[TARGET_COL] >= 0
]

df = df[
    df[FEATURE_COL].isna() |
    (df[FEATURE_COL] >= 0)
]

df = df.reset_index(drop=True)

if len(df) < 12:
    raise ValueError(
        "Insufficient records for PSO training."
    )

print("\nCleaned Dataset:")
print(df.head())

print("\nTotal Records:", len(df))
print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 5. FEATURES AND TARGET
# ============================================================

X = df[[FEATURE_COL]]
y = df[TARGET_COL]

indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.25,
    random_state=SEED
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

test_states = df.iloc[
    test_idx
][STATE_COL].reset_index(drop=True)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

# ============================================================
# 6. PARTICLE SWARM OPTIMIZATION
# ============================================================

print("\n========== PSO OPTIMIZATION ==========")

# PSO minimizes a fitness function.
# Each particle represents model hyperparameters.

class ParticleSwarmOptimizer:

    def __init__(
        self,
        fitness_function,
        bounds,
        particles=8,
        iterations=12,
        inertia=0.7,
        cognitive=1.5,
        social=1.5,
        seed=SEED
    ):

        self.fitness_function = fitness_function

        self.bounds = np.array(
            bounds,
            dtype=float
        )

        self.particles = particles
        self.iterations = iterations

        self.inertia = inertia
        self.cognitive = cognitive
        self.social = social

        self.rng = np.random.default_rng(seed)

        self.dimensions = len(bounds)

        self.lower = self.bounds[:, 0]
        self.upper = self.bounds[:, 1]

        self.positions = self.rng.uniform(
            self.lower,
            self.upper,
            size=(particles, self.dimensions)
        )

        self.velocities = self.rng.uniform(
            -0.1,
            0.1,
            size=(particles, self.dimensions)
        ) * (self.upper - self.lower)

        self.personal_best_positions = (
            self.positions.copy()
        )

        self.personal_best_scores = np.full(
            particles,
            np.inf
        )

        self.global_best_position = None
        self.global_best_score = np.inf

        self.history = []

    def optimize(self):

        for iteration in range(
            1,
            self.iterations + 1
        ):

            # Evaluate particles
            for i in range(self.particles):

                score = self.fitness_function(
                    self.positions[i]
                )

                if score < self.personal_best_scores[i]:

                    self.personal_best_scores[i] = score

                    self.personal_best_positions[i] = (
                        self.positions[i].copy()
                    )

                if score < self.global_best_score:

                    self.global_best_score = score

                    self.global_best_position = (
                        self.positions[i].copy()
                    )

            # Update particle velocities and positions
            for i in range(self.particles):

                r1 = self.rng.random(
                    self.dimensions
                )

                r2 = self.rng.random(
                    self.dimensions
                )

                cognitive_velocity = (
                    self.cognitive *
                    r1 *
                    (
                        self.personal_best_positions[i] -
                        self.positions[i]
                    )
                )

                social_velocity = (
                    self.social *
                    r2 *
                    (
                        self.global_best_position -
                        self.positions[i]
                    )
                )

                self.velocities[i] = (
                    self.inertia * self.velocities[i] +
                    cognitive_velocity +
                    social_velocity
                )

                # Limit maximum velocity
                max_velocity = (
                    0.2 * (self.upper - self.lower)
                )

                self.velocities[i] = np.clip(
                    self.velocities[i],
                    -max_velocity,
                    max_velocity
                )

                self.positions[i] += (
                    self.velocities[i]
                )

                self.positions[i] = np.clip(
                    self.positions[i],
                    self.lower,
                    self.upper
                )

            self.history.append({
                "Iteration": iteration,
                "Best_Fitness": float(
                    self.global_best_score
                )
            })

            print(
                f"Iteration {iteration}/{self.iterations}"
                f" | Best RMSE: "
                f"{self.global_best_score:.4f}"
            )

        return (
            self.global_best_position,
            self.global_best_score,
            pd.DataFrame(self.history)
        )


# ============================================================
# 7. PSO RANDOM FOREST SEARCH SPACE
# ============================================================

# Particle dimensions:
# n_estimators
# max_depth
# min_samples_leaf
# min_samples_split
# max_features

RF_BOUNDS = [
    (30, 180),
    (2, 12),
    (1, 5),
    (2, 10),
    (0.5, 1.0)
]


def rf_particle_to_params(particle):

    return {
        "n_estimators": int(
            np.clip(round(particle[0]), 30, 180)
        ),
        "max_depth": int(
            np.clip(round(particle[1]), 2, 12)
        ),
        "min_samples_leaf": int(
            np.clip(round(particle[2]), 1, 5)
        ),
        "min_samples_split": int(
            np.clip(round(particle[3]), 2, 10)
        ),
        "max_features": float(
            np.clip(particle[4], 0.5, 1.0)
        )
    }


def create_rf_pipeline(params):

    return Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "model",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=1
            )
        )
    ])


# ============================================================
# 8. RANDOM FOREST FITNESS FUNCTION
# ============================================================

cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

rf_fitness_cache = {}


def rf_fitness(particle):

    params = rf_particle_to_params(
        particle
    )

    key = (
        params["n_estimators"],
        params["max_depth"],
        params["min_samples_leaf"],
        params["min_samples_split"],
        round(params["max_features"], 5)
    )

    if key in rf_fitness_cache:
        return rf_fitness_cache[key]

    model = create_rf_pipeline(
        params
    )

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_squared_error",
        n_jobs=1
    )

    rmse = np.sqrt(
        np.maximum(-scores, 0)
    ).mean()

    rf_fitness_cache[key] = float(rmse)

    return float(rmse)


# ============================================================
# 9. RUN PSO FOR RANDOM FOREST
# ============================================================

print("\n========== PSO RANDOM FOREST ==========")

rf_optimizer = ParticleSwarmOptimizer(
    fitness_function=rf_fitness,
    bounds=RF_BOUNDS,
    particles=8,
    iterations=12,
    inertia=0.7,
    cognitive=1.5,
    social=1.5
)

best_rf_particle, best_rf_cv_rmse, rf_history = (
    rf_optimizer.optimize()
)

best_rf_params = rf_particle_to_params(
    best_rf_particle
)

print("\nBest RF Parameters:")
print(best_rf_params)

print("\nBest Cross-Validation RMSE:")
print(best_rf_cv_rmse)

# ============================================================
# 10. TRAIN FINAL PSO RANDOM FOREST
# ============================================================

pso_rf = create_rf_pipeline(
    best_rf_params
)

pso_rf.fit(
    X_train,
    y_train
)

rf_predictions = pso_rf.predict(
    X_test
)

rf_predictions = np.maximum(
    rf_predictions,
    0
)

rf_metrics = calculate_metrics(
    y_test,
    rf_predictions
)

print("\nPSO Random Forest Results:")
print(rf_metrics)

# Save PKL
with open(
    output_path("model.pkl"),
    "wb"
) as file:
    pickle.dump(
        pso_rf,
        file
    )

print("\nSaved:", output_path("model.pkl"))

# ============================================================
# 11. PSO NEURAL NETWORK SEARCH SPACE
# ============================================================

print("\n========== PSO NEURAL NETWORK ==========")

# Particle dimensions:
# first hidden layer units
# second hidden layer units
# learning rate

NN_BOUNDS = [
    (8, 64),
    (4, 32),
    (0.0003, 0.01)
]


def nn_particle_to_params(particle):

    return [
        int(
            np.clip(
                round(particle[0]),
                8,
                64
            )
        ),
        int(
            np.clip(
                round(particle[1]),
                4,
                32
            )
        ),
        float(
            np.clip(
                particle[2],
                0.0003,
                0.01
            )
        )
    ]


def build_neural_network(params):

    units1, units2, learning_rate = params

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(1,)),

        tf.keras.layers.Dense(
            units1,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            units2,
            activation="relu"
        ),

        tf.keras.layers.Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="mse",
        metrics=["mae"]
    )

    return model


# ============================================================
# 12. NEURAL NETWORK VALIDATION DATA
# ============================================================

nn_train_idx, nn_val_idx = train_test_split(
    np.arange(len(X_train)),
    test_size=0.25,
    random_state=SEED
)

X_nn_train = X_train.iloc[
    nn_train_idx
]

X_nn_val = X_train.iloc[
    nn_val_idx
]

y_nn_train = y_train.iloc[
    nn_train_idx
]

y_nn_val = y_train.iloc[
    nn_val_idx
]

search_imputer = SimpleImputer(
    strategy="median"
)

search_feature_scaler = StandardScaler()
search_target_scaler = StandardScaler()

X_nn_train_imp = search_imputer.fit_transform(
    X_nn_train
)

X_nn_val_imp = search_imputer.transform(
    X_nn_val
)

X_nn_train_scaled = (
    search_feature_scaler.fit_transform(
        X_nn_train_imp
    )
)

X_nn_val_scaled = (
    search_feature_scaler.transform(
        X_nn_val_imp
    )
)

y_nn_train_scaled = (
    search_target_scaler.fit_transform(
        y_nn_train.to_numpy().reshape(-1, 1)
    )
)

y_nn_val_scaled = (
    search_target_scaler.transform(
        y_nn_val.to_numpy().reshape(-1, 1)
    )
)

# ============================================================
# 13. NEURAL NETWORK FITNESS FUNCTION
# ============================================================

nn_fitness_cache = {}


def nn_fitness(particle):

    params = nn_particle_to_params(
        particle
    )

    key = (
        params[0],
        params[1],
        round(params[2], 7)
    )

    if key in nn_fitness_cache:
        return nn_fitness_cache[key]

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(SEED)

    model = build_neural_network(
        params
    )

    early_stopping = (
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=12,
            restore_best_weights=True
        )
    )

    model.fit(
        X_nn_train_scaled,
        y_nn_train_scaled,
        validation_data=(
            X_nn_val_scaled,
            y_nn_val_scaled
        ),
        epochs=100,
        batch_size=4,
        callbacks=[early_stopping],
        verbose=0
    )

    pred_scaled = model.predict(
        X_nn_val_scaled,
        verbose=0
    )

    predictions = (
        search_target_scaler.inverse_transform(
            pred_scaled
        ).flatten()
    )

    predictions = np.maximum(
        predictions,
        0
    )

    score = np.sqrt(
        mean_squared_error(
            y_nn_val,
            predictions
        )
    )

    nn_fitness_cache[key] = float(score)

    return float(score)


# ============================================================
# 14. RUN PSO FOR NEURAL NETWORK
# ============================================================

nn_optimizer = ParticleSwarmOptimizer(
    fitness_function=nn_fitness,
    bounds=NN_BOUNDS,
    particles=5,
    iterations=5,
    inertia=0.7,
    cognitive=1.5,
    social=1.5,
    seed=SEED + 1
)

best_nn_particle, best_nn_val_rmse, nn_history = (
    nn_optimizer.optimize()
)

best_nn_params = nn_particle_to_params(
    best_nn_particle
)

print("\nBest Neural Network Parameters:")
print(best_nn_params)

print("\nBest Validation RMSE:")
print(best_nn_val_rmse)

# ============================================================
# 15. TRAIN FINAL PSO NEURAL NETWORK
# ============================================================

final_imputer = SimpleImputer(
    strategy="median"
)

final_feature_scaler = StandardScaler()
final_target_scaler = StandardScaler()

X_train_imp = final_imputer.fit_transform(
    X_train
)

X_test_imp = final_imputer.transform(
    X_test
)

X_train_scaled = (
    final_feature_scaler.fit_transform(
        X_train_imp
    )
)

X_test_scaled = (
    final_feature_scaler.transform(
        X_test_imp
    )
)

y_train_scaled = (
    final_target_scaler.fit_transform(
        y_train.to_numpy().reshape(-1, 1)
    )
)

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

pso_nn = build_neural_network(
    best_nn_params
)

final_early_stopping = (
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=20,
        restore_best_weights=True
    )
)

final_nn_history = pso_nn.fit(
    X_train_scaled,
    y_train_scaled,
    validation_split=0.2,
    epochs=250,
    batch_size=4,
    callbacks=[final_early_stopping],
    verbose=0
)

nn_pred_scaled = pso_nn.predict(
    X_test_scaled,
    verbose=0
)

nn_predictions = (
    final_target_scaler.inverse_transform(
        nn_pred_scaled
    ).flatten()
)

nn_predictions = np.maximum(
    nn_predictions,
    0
)

nn_metrics = calculate_metrics(
    y_test,
    nn_predictions
)

print("\nPSO Neural Network Results:")
print(nn_metrics)

# Save H5 model
pso_nn.save(
    output_path("model.h5")
)

print("\nSaved:", output_path("model.h5"))

# ============================================================
# 16. SAVE RESULT CSV
# ============================================================

print("\n========== RESULT CSV ==========")

result_df = pd.DataFrame([
    {
        "Model": "PSO Random Forest",
        **rf_metrics
    },
    {
        "Model": "PSO Neural Network",
        **nn_metrics
    }
])

result_df = result_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print(result_df.to_string(index=False))

# ============================================================
# 17. SAVE PREDICTION CSV
# ============================================================

print("\n========== PREDICTION CSV ==========")

prediction_df = pd.DataFrame({
    "State/UT": test_states,
    "Seller_Points": X_test[
        FEATURE_COL
    ].to_numpy(),
    "Actual_Product_Listings": y_test.to_numpy(),
    "PSO_RF_Prediction": np.round(
        rf_predictions,
        2
    ),
    "PSO_NN_Prediction": np.round(
        nn_predictions,
        2
    )
})

prediction_df = prediction_df.sort_values(
    "Actual_Product_Listings",
    ascending=False
).reset_index(drop=True)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print(prediction_df.to_string(index=False))

# ============================================================
# 18. SAVE OPTIMIZATION HISTORY CSV
# ============================================================

rf_history.to_csv(
    output_path("optimization_history.csv"),
    index=False
)

nn_history.to_csv(
    output_path("nn_optimization_history.csv"),
    index=False
)

# ============================================================
# 19. PSO ACCURACY GRAPH
# ============================================================

print("\n========== ACCURACY GRAPH ==========")

plt.figure(figsize=(10, 6))

bars = plt.bar(
    result_df["Model"],
    result_df["R2_Percentage"],
    color=["#277DA1", "#F8961E"]
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

for bar, value in zip(
    bars,
    result_df["R2_Percentage"]
):

    plt.annotate(
        f"{value:.2f}%",
        (
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        ha="center",
        va="bottom" if value >= 0 else "top",
        fontsize=11
    )

plt.title(
    "PSO Model Performance - R2 Score",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("PSO Models")
plt.ylabel("R2 Score (%)")
plt.grid(axis="y", alpha=0.3)

save_show("accuracy_graph.png")

# ============================================================
# 20. PSO PREDICTION GRAPH
# ============================================================

print("\n========== PREDICTION GRAPH ==========")

best_model_name = result_df.iloc[0]["Model"]

best_prediction_col = (
    "PSO_RF_Prediction"
    if best_model_name == "PSO Random Forest"
    else "PSO_NN_Prediction"
)

x_axis = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(13, 7))

plt.plot(
    x_axis,
    prediction_df["Actual_Product_Listings"],
    marker="o",
    linewidth=2.5,
    label="Actual Product Listings"
)

plt.plot(
    x_axis,
    prediction_df[best_prediction_col],
    marker="s",
    linestyle="--",
    linewidth=2,
    label=best_model_name
)

plt.xticks(
    x_axis,
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.title(
    "PSO Actual vs Predicted Product Listings",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Number of Product Listings")
plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ============================================================
# 21. PSO RESULT GRAPH
# ============================================================

print("\n========== RESULT GRAPH ==========")

x_axis = np.arange(
    len(result_df)
)

width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x_axis - width / 2,
    result_df["MAE"],
    width,
    label="MAE"
)

plt.bar(
    x_axis + width / 2,
    result_df["RMSE"],
    width,
    label="RMSE"
)

plt.xticks(
    x_axis,
    result_df["Model"]
)

plt.title(
    "PSO Model Error Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Error (Product Listings)")
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 22. PSO COMPARISON GRAPH
# ============================================================

print("\n========== COMPARISON GRAPH ==========")

x_axis = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(14, 8))

plt.plot(
    x_axis,
    prediction_df["Actual_Product_Listings"],
    color="black",
    marker="o",
    linewidth=3,
    label="Actual"
)

plt.plot(
    x_axis,
    prediction_df["PSO_RF_Prediction"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Random Forest"
)

plt.plot(
    x_axis,
    prediction_df["PSO_NN_Prediction"],
    marker="^",
    linestyle="--",
    linewidth=2,
    label="PSO Neural Network"
)

plt.xticks(
    x_axis,
    prediction_df["State/UT"],
    rotation=45,
    ha="right"
)

plt.title(
    "PSO Model Prediction Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Number of Product Listings")
plt.legend()
plt.grid(alpha=0.3)

save_show("comparison_graph.png")

# ============================================================
# 23. PSO CORRELATION HEATMAP
# ============================================================

print("\n========== HEATMAP ==========")

heatmap_df = df[
    [FEATURE_COL, TARGET_COL]
].copy()

heatmap_df["Products_Per_Seller"] = np.where(
    heatmap_df[FEATURE_COL] > 0,
    heatmap_df[TARGET_COL] /
    heatmap_df[FEATURE_COL],
    np.nan
)

correlation = heatmap_df.corr(
    method="pearson"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0,
    square=True
)

plt.title(
    "PSO Project - Dataset Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

save_show("heatmap.png")

# ============================================================
# 24. PSO RANDOM FOREST FITNESS GRAPH
# ============================================================

print("\n========== PSO FITNESS GRAPH ==========")

plt.figure(figsize=(11, 6))

plt.plot(
    rf_history["Iteration"],
    rf_history["Best_Fitness"],
    marker="o",
    linewidth=2.5,
    label="PSO Random Forest"
)

plt.title(
    "PSO Random Forest Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("PSO Iteration")
plt.ylabel("Best Cross-Validation RMSE")
plt.legend()
plt.grid(alpha=0.3)

save_show("fitness_graph.png")

# ============================================================
# 25. PSO NEURAL NETWORK FITNESS GRAPH
# ============================================================

print("\n========== PSO NN FITNESS GRAPH ==========")

plt.figure(figsize=(11, 6))

plt.plot(
    nn_history["Iteration"],
    nn_history["Best_Fitness"],
    marker="o",
    linewidth=2.5,
    label="PSO Neural Network"
)

plt.title(
    "PSO Neural Network Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("PSO Iteration")
plt.ylabel("Best Validation RMSE")
plt.legend()
plt.grid(alpha=0.3)

save_show("nn_fitness_graph.png")

# ============================================================
# 26. SAVE YAML CONFIGURATION
# ============================================================

print("\n========== SAVING YAML ==========")

config = {
    "project": (
        "AI-Based State-Wise Seller Network "
        "and Product Availability Analysis and Prediction"
    ),
    "algorithm": "Particle Swarm Optimization",
    "task": "Regression",
    "dataset": CSV_PATH,
    "feature": FEATURE_COL,
    "target": TARGET_COL,
    "random_state": SEED,
    "test_size": 0.25,
    "pso_random_forest": {
        "particles": 8,
        "iterations": 12,
        "inertia": 0.7,
        "cognitive": 1.5,
        "social": 1.5,
        "best_parameters": best_rf_params,
        "best_cv_rmse": float(best_rf_cv_rmse)
    },
    "pso_neural_network": {
        "particles": 5,
        "iterations": 5,
        "inertia": 0.7,
        "cognitive": 1.5,
        "social": 1.5,
        "best_parameters": {
            "units1": best_nn_params[0],
            "units2": best_nn_params[1],
            "learning_rate": best_nn_params[2]
        },
        "best_validation_rmse": float(
            best_nn_val_rmse
        )
    },
    "output_directory": BASE_DIR
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", output_path("config.yaml"))

# ============================================================
# 27. SAVE JSON METADATA
# ============================================================

print("\n========== SAVING JSON ==========")

generated_files = [
    "pso_model.pkl",
    "pso_model.h5",
    "pso_config.yaml",
    "pso_metadata.json",
    "pso_accuracy_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_comparison_graph.png",
    "pso_heatmap.png",
    "pso_fitness_graph.png",
    "pso_nn_fitness_graph.png",
    "pso_optimization_history.csv",
    "pso_nn_optimization_history.csv"
]

metadata = {
    "project": config["project"],
    "algorithm": "Particle Swarm Optimization",
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),
    "total_seller_points": float(
        df[FEATURE_COL].sum()
    ),
    "total_product_listings": float(
        df[TARGET_COL].sum()
    ),
    "feature": FEATURE_COL,
    "target": TARGET_COL,
    "pso_rf_best_parameters": best_rf_params,
    "pso_rf_best_cv_rmse": float(best_rf_cv_rmse),
    "pso_nn_best_parameters": {
        "units1": best_nn_params[0],
        "units2": best_nn_params[1],
        "learning_rate": best_nn_params[2]
    },
    "pso_nn_best_validation_rmse": float(
        best_nn_val_rmse
    ),
    "model_results": result_df.to_dict(
        orient="records"
    ),
    "neural_network_preprocessing": {
        "imputer_statistics": (
            final_imputer.statistics_.tolist()
        ),
        "feature_mean": (
            final_feature_scaler.mean_.tolist()
        ),
        "feature_scale": (
            final_feature_scaler.scale_.tolist()
        ),
        "target_mean": (
            final_target_scaler.mean_.tolist()
        ),
        "target_scale": (
            final_target_scaler.scale_.tolist()
        )
    },
    "generated_files": generated_files
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Saved:", output_path("metadata.json"))

# ============================================================
# 28. FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("==========================================")

print("\nPSO RANDOM FOREST RESULTS:")
print(rf_metrics)

print("\nPSO NEURAL NETWORK RESULTS:")
print(nn_metrics)

print("\nBEST MODEL BY TEST RMSE:")
print(best_model_name)

print("\nGENERATED FILES:")

for filename in generated_files:

    full_path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(full_path):
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOUTPUT DIRECTORY:")
print(BASE_DIR)

print("\nAll PSO results generated successfully!")
```
